## Transformación negativa

s = L - 1 ｰ r

s = 256 - 1 - r

s = 255 - r

```python
plt.imshow(canal_b, cmap='gray')
plt.show()
```

```python
filas, columnas = canal_b.shape

# Faltó crear una copia de la imagen para no dañar la original

for f in range(filas):
    for c in range(columnas):
        canal_b[f, c] = 255 - canal_b[f, c]

plt.imshow(canal_b, cmap='gray')
plt.title("Canal_b negativo")
plt.show()
```

```python
# Umbralización

filas, columnas = canal_g.shape
k = 200

# Faltó crear una copia de la imagen para no dañar la original

for f in range(filas):
    for c in range(columnas):
        if canal_g[f, c] < k:
            canal_g[f, c] = 0
        else:
            canal_g[f, c] = 255

plt.imshow(canal_g, cmap='gray')
plt.title("Canal_g umbralizado")
plt.show()
```

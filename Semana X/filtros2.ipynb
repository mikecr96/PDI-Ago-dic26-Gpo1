```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv.imread("/Users/miguelcamargorojas/iCloud Drive (Archive)/Documents/UP/PDI-Ago-dic26/PDI-Ago-dic26-Gpo1/images/aloeR.jpg", 0)
plt.imshow(img, cmap='gray')
plt.show()
```

```python
kernel = np.ones((3, 3)) * 1/9
kernel.sum()
```

```python
img_filtrada = cv.filter2D(img, -1, kernel)
```

```python
np.unique(img_filtrada)
```

```python
plt.imshow(np.hstack((img, img_filtrada)), cmap='gray')
plt.show()
```

```python
img == img_filtrada
```

```python
def getKernel(n:int):
    return np.ones((n, n)) * 1/n**2
```

```python
img_filtrada = cv.filter2D(img, -1, getKernel(21))
plt.imshow(np.hstack((img, img_filtrada)), cmap='gray')
plt.show()
```

## Filtro Laplaciano

```python
laplaciano1 = np.array([
    [0, -1, 0],
    [-1, 4, -1],
    [0, -1, 0]
])

laplaciano2 = laplaciano1 * -1

laplaciano2.sum()
```

```python
img_filtrada2 = cv.filter2D(img, -1, laplaciano1)

plt.imshow(np.hstack((img, img_filtrada2)), cmap='gray')
plt.title("Filtro con laplaciano 1")
plt.show()
```

```python
np.unique(img_filtrada2)
```

```python
img_filtrada3 = cv.filter2D(img, -1, laplaciano2)

plt.imshow(np.hstack((img, img_filtrada3)), cmap='gray')
plt.title("Filtro con laplaciano 2")
plt.show()
```

```python
laplaciano3 = np.array([
    [-1, -1, -1],
    [-1, 8, -1],
    [-1, -1, -1]
])
laplaciano4 = laplaciano3 * -1

laplaciano4.sum()
```

```python
img_filtrada4 = cv.filter2D(img, -1, laplaciano3)

plt.imshow(np.hstack((img, img_filtrada4)), cmap='gray')
plt.title("Filtro con laplaciano 3")
plt.show()
```

```python
img_filtrada5 = cv.filter2D(img, -1, laplaciano4)

plt.imshow(np.hstack((img, img_filtrada5)), cmap='gray')
plt.title("Filtro con laplaciano 4")
plt.show()
```

```python
plt.imshow(cv.Laplacian(img, -1), cmap='gray')
plt.show()
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

# Umbralización GLOBAL

```python
sudoku_gray = cv.imread('/Users/miguelcamargorojas/iCloud Drive (Archive)/Documents/UP/PDI-Ago-dic26/PDI-Ago-dic26-Gpo1/images/sudoku.png', 0)
plt.imshow(sudoku_gray, 'gray')
plt.show()
```

```python
def mostrar_img(img, title='Título genérico'):
    plt.imshow(img, 'gray')
    plt.title(title)
    plt.axis('off')
    plt.show()
```

```python
k = 50
sudoku_umb = cv.threshold(sudoku_gray, k, 255, cv.THRESH_BINARY)[1]
mostrar_img(sudoku_umb, f'Imagen umbralizada con umbral {k}')
```

```python
umb, sudoku_umb = cv.threshold(sudoku_gray, 0, 255, cv.THRESH_BINARY | cv.THRESH_OTSU)
mostrar_img(sudoku_umb, f'Imagen umbralizada con Otsu, umbral {int(umb)}')
```

```python
umb, sudoku_umb = cv.threshold(sudoku_gray, 100000, 255, cv.THRESH_BINARY | cv.THRESH_TRIANGLE)
mostrar_img(sudoku_umb, f'Imagen umbralizada con Triangle, umbral {int(umb)}')
```

```python
promedio = int(np.mean(sudoku_gray))
umb, sudoku_umb = cv.threshold(sudoku_gray, promedio, 255, cv.THRESH_BINARY)
mostrar_img(sudoku_umb, f'Imagen umbralizada con umbral promedio {promedio}')
```

# Umbralización Adaptativa

```python
sudoku_gray_blur = cv.medianBlur(sudoku_gray, 5)
sudoku_umb_mean = cv.adaptiveThreshold(sudoku_gray_blur, 255, cv.ADAPTIVE_THRESH_MEAN_C, cv.THRESH_BINARY, 11, 2)
mostrar_img(sudoku_umb_mean)
```

```python
# creen una imagen de 100 x 100 totalmente negra y muestrenla

img_negra = np.zeros((100, 100), dtype=np.uint8)
plt.imshow(img_negra, cmap="gray")
plt.show()
```

```python
# Slicing
# creen una imagen de 100 x 100 totalmente negra y muestrenla

img_negra = np.zeros((99, 99), dtype=np.uint8)
img_negra[:34] = 127
img_negra[66:] = 255
plt.imshow(img_negra.T, cmap="gray")
plt.show()
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
cosas = cv.imread('/Users/miguelcamargorojas/iCloud Drive (Archive)/Documents/UP/PDI-Ago-dic26/PDI-Ago-dic26-Gpo1/images/stuff.jpg', 0)
plt.imshow(cosas, 'gray')
plt.show()
```

```python
def getKernel(n:int):
    return np.ones((n, n)) * 1/(n*n)

getKernel(3)
```

```python
cosas_filtrada = cv.filter2D(cosas, -1, getKernel(11))
plt.imshow(cosas_filtrada, 'gray')
plt.show()
```

```python
kernel1 = np.array([
    [0, -1, 0],
    [-1, 4, -1],
    [0, -1, 0]
])
```

```python
cosas_pasa_altas = cv.filter2D(cosas, -1, kernel1)
plt.imshow(cosas_pasa_altas, 'gray')
plt.show()
```

```python
colores_bgr = ['blue', 'green', 'red']
colores_rgb = ['red', 'green', 'blue']
for i in range(3):
    histograma = cv.calcHist([cosas_rgb], [i], None, [256], [0, 256])
    plt.bar(range(256), histograma, color=colores_rgb[i])
plt.show()
```

```python
"""
1) cargar imagen en escala de grises
2) filtrar con desenfoque Gaussiano la imagen
3) al resultado calcular el laplaciano con cv.Laplacian
4) mostrar el resultado
"""
# 1 lo omito ya la cargué
# 2
cosas_gauss = cv.GaussianBlur(cosas, (3, 3), 2)
# 3
cosas_gauss_laplaciano = cv.Laplacian(cosas_gauss, -1)
# 4
plt.imshow(cosas_gauss_laplaciano, 'gray')
plt.show()
```

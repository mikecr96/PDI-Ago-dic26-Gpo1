## Ecualizar el histograma

```python
imagen_eq = cv.equalizeHist(imagen)
```

```python
plt.imshow(np.hstack((imagen, imagen_eq)), cmap='gray')
plt.show()
```

```python
histograma_eq = cv.calcHist([imagen_eq], (0,), None, [256], (0, 256))
plt.bar(range(256), histograma_eq)
plt.show()
```

```python
# La bandera 0, carga la imagen en escala de grises
siames_bgr = cv.imread(r"/Users/miguelcamargorojas/Downloads/siames.jpg")
plt.imshow(siames_bgr[:, :, ::-1])
plt.show()
```

```python
histograma_b = cv.calcHist([siames_bgr], (0,), None, [256], (0, 256))
plt.bar(range(256), histograma_b)
plt.title('Histograma del canal azul')
plt.show()
```

```python
histograma_g = cv.calcHist([siames_bgr], (1,), None, [256], (0, 256))
plt.bar(range(256), histograma_g)
plt.title('Histograma del canal verde')
plt.show()
```

```python
histograma_r = cv.calcHist([siames_bgr], (2,), None, [256], (0, 256))
plt.bar(range(256), histograma_r)
plt.title('Histograma del canal rojo')
plt.show()
```

```python
for c in range(3):
    histograma = cv.calcHist([siames_bgr], (c,), None, [256], (0, 256))
    plt.bar(range(256), histograma)
    plt.title(f'Histograma del canal {c}')
plt.show()
```

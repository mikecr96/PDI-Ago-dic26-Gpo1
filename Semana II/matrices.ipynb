```python
# Visite todos los elementos de una matriz 200x200 y a c/u de los elementos le sume 10
mat2 = np.random.randint(10, size=(200, 200))
print(mat2)

fila, columna = mat2.shape

for i in range(fila): # Filas
    for j in range(columna): # Columnas
        mat2[i, j] += 10

mat2
```

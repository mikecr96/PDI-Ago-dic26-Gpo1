## Convertir a escala de grises

```python
plt.imshow(imagen[:, :, 0], cmap='gray')
plt.show()
```

```python
# Separar canales con sintaxis pura de Python
canal_R = imagen[:, :, 0]
canal_G = imagen[:, :, 1]
canal_B = imagen[:, :, 2]

# sumar los 3 canales y dividir entre 3 (promedio)
imagen_gris = (canal_R + canal_G + canal_B) / 3
imagen_gris = imagen_gris.astype(np.uint8)
```

```python
plt.imshow(imagen_gris, cmap='gray')
plt.show()
```

## Aislamiento cromático

```python
# Muestre la imagen que cargó a color, solo con su componente rojo
# (Muestre la imagen en rojo)
print(imagen.shape)

# Apagar los canales G,B
# 1. Crear una copia para no afectar a la imagen original
imagen_roja = imagen.copy()

# 2. Apagar los canales G,B
imagen_roja[:, :, 1] = 0
imagen_roja[:, :, 2] = 0

plt.imshow(imagen_roja)
plt.show()
```

```python
# Separar componente cromático azul y verde

# Muestre la imagen que cargó a color, solo con su componente verde
# (Muestre la imagen en verde)

# Apagar los canales R,B
# 1. Crear una copia para no afectar a la imagen original
imagen_verde = imagen.copy()

# 2. Apagar los canales G,B
imagen_verde[:, :, (0, 2)] = 0

plt.imshow(imagen_verde)
plt.show()
```

## Mezcla de canales

```python
# R + G (Sin azul) -> amarillo

# Apagar canal B
# 1. Crear una copia para no afectar a la imagen original
imagen_amarilla = imagen.copy()

# 2. Apagar los canales G,B
imagen_amarilla[:, :, 2] = 0

plt.imshow(imagen_amarilla)
plt.show()
```

```python
# R + B (Sin verde) -> magenta

# Apagar canal G
# 1. Crear una copia para no afectar a la imagen original
imagen_magenta = imagen.copy()

# 2. Apagar los canales G,B
imagen_magenta[:, :, 1] = 0

plt.imshow(imagen_magenta)
plt.show()
```

```python
# G + B (Sin rojo) -> cian

# Apagar canal R
# 1. Crear una copia para no afectar a la imagen original
imagen_cian = imagen.copy()

# 2. Apagar los canales G,B
imagen_cian[:, :, 0] = 0

plt.imshow(imagen_cian)
plt.show()
```

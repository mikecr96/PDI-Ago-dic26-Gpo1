## Convertir a escala de grises

```python
plt.imshow(imagen[:, :, 0], cmap='gray')
plt.show()
```

```python
# Separar canales con sintaxis pura de Python
canal_R = imagen[:, :, 0]
canal_G = imagen[:, :, 1]
canal_B = imagen[:, :, 2]

# sumar los 3 canales y dividir entre 3 (promedio)
imagen_gris = (canal_R + canal_G + canal_B) / 3
imagen_gris = imagen_gris.astype(np.uint8)
```

```python
plt.imshow(imagen_gris, cmap='gray')
plt.show()
```
